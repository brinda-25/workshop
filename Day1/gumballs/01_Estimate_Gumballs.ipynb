## Download Data

```python
RAW_ESTIMATES = 'https://docs.google.com/spreadsheets/d/1b8TyWWb2LMYE3_3vAWayTb3tDttOAlCvYg3OQuKBZVI/export?format=csv'
```

```python
df = pd.read_csv(RAW_ESTIMATES)
df.head(10)
```

## Review Data

### How many respondents ?

```python
df.index.size
```

```python
df['Email Address'].nunique()
```

### Review Data Distribution

#### What is the Minimum guess ?

```python
df['How many gumballs do you see ?'].min().astype(int).item()
```

#### What is the Maximum guess ?

```python
df['How many gumballs do you see ?'].max().astype(int).item()
```

#### What is the most popular guess ? 

```python
df['How many gumballs do you see ?'].value_counts().head(1)
```

### Visualize the data distribution

```python
sns.kdeplot(df['How many gumballs do you see ?'], clip=[0, None], cut=0, bw_adjust=0.5, fill=True)
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

#### What is the Average / Mean ?

```python
df['How many gumballs do you see ?'].mean().astype(int).item()
```
