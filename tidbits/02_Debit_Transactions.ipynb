## What an unrealistic distribution might look like ?

```python
def plot_unrealistic_distribution():
    index = list(range(1,10))
    counts = np.random.choice(index, size=9)
    dist = counts/counts.sum()
    pd.Series(data=dist, index=index).plot.bar()
```

```python
plot_unrealistic_distribution()
```
