# Visualizing Functions

```python
import pandas as pd
import numpy as np
import math
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

## Define functions

```python
def f1(x, a, b):
    return a*x + b

def f2(x, a, b, c):
    return a*(x**2) + b*x + c

def f3(x, a, b, c, d):
    return a*(x**3) + b*(x**2) + c*x + d

def f4(x, a):
    return math.exp(a*x)

def f5(x, a):
    return math.log(a*x)

def f6(x, a):
    return math.sin(a*x)
```

### Helpers to plot function

```python
def plot_f1(display_equation = False):
    a = 1
    b = 0
    equation = f'y = {a}x + {b}'
    equation = 'y = x'
    xy = np.array([(i, f1(i, a, b)) for i in range(-10, 11)])
    plt.plot(xy[:,0], xy[:,1])
    if display_equation:
        plt.title(equation)

def plot_f2(display_equation = False):
    a = 1
    b = 0
    c = 0
    equation = f'y = {a}x^2 + {b}x + {c}'
    equation = 'y = x^2'
    xy = np.array([(i, f2(i, a, b, c)) for i in range(-10, 11)])
    plt.plot(xy[:,0], xy[:,1])
    if display_equation:
        plt.title(equation)

def plot_f3(display_equation = False):
    a = 1
    b = 0
    c = 0
    d = 0
    equation = f'y = {a}x^3 + {b}x^2 + {c}x + {d}'
    equation = 'y = x^3'
    xy = np.array([(i, f3(i, a, b, c, d)) for i in range(-10, 11)])
    plt.plot(xy[:,0], xy[:,1])
    if display_equation:
        plt.title(equation)

def plot_f4(display_equation = False):
    a = 1
    equation = f'y = exp({a}x)'
    equation = f'y = exp(x)'
    xy = np.array([(i, f4(i, a)) for i in np.linspace(6, 15)])
    plt.plot(xy[:,0], xy[:,1])
    if display_equation:
        plt.title(equation)

def plot_f5(display_equation = False):
    a = 1
    equation = f'y = log({a}x)'
    equation = f'y = log(x)'
    xy = np.array([(i, f5(i, a)) for i in np.linspace(1, 1000)])
    plt.plot(xy[:,0], xy[:,1])
    if display_equation:
        plt.title(equation)

def plot_f6(display_equation = False):
    a = 1
    equation = f'y = sin({a}x)'
    equation = f'y = sin(x)'
    xy = np.array([(i, f6(i, a)) for i in np.linspace(-10, 11, num=100)])
    plt.plot(xy[:,0], xy[:,1])
    if display_equation:
        plt.title(equation)

def plot_combo1(display_equation = False):
    a = 1
    b = 0
    c = 1
    equation = f'y = {a}x + sin({c}x)+ b'
    equation = f'y = x + sin(x)'
    xy = np.array([(i, f1(i, a, b) + f6(i, c)) for i in np.linspace(-10, 11, num=100)])
    plt.plot(xy[:,0], xy[:,1])
    if display_equation:
        plt.title(equation)
```

## Let us Visualize the functions

```python
plot_f1(display_equation=False)
```
